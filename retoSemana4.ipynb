```python
#6
import matplotlib.pyplot as plt

def graficar_recta(x1, y1, x2, y2):

    m = (y2 - y1) / (x2 - x1)


    print(f"La pendiente de la recta es: {m}")


    x_vals = [x1, x2]
    y_vals = [y1, y2]


    plt.plot(x_vals, y_vals, marker='o', linestyle='-', color='purple')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('Gráfica de una recta')
    plt.grid(True)
    plt.show()


graficar_recta(1, 2, 4, 6)
```

```python
#7
import numpy as np
import matplotlib.pyplot as plt

def movimiento_parabolico(v0, angulo):

    angulo = np.radians(angulo)


    g = 9.81


    v0x = v0 * np.cos(angulo)
    v0y = v0 * np.sin(angulo)


    t_vuelo = (2 * v0y) / g


    t = np.linspace(0, t_vuelo, 300)
    x = v0x * t
    y = v0y * t - 0.5 * g * t**2


    plt.plot(x, y, label=f'v0={v0} m/s, ángulo={np.degrees(angulo)}°')
    plt.xlabel('Distancia (m)')
    plt.ylabel('Altura (m)')
    plt.title('Movimiento Parabólico')
    plt.legend()
    plt.grid(True)
    plt.show()

# Ejemplo de uso
movimiento_parabolico(50, 45)
```

```python
#8
import numpy as np
import matplotlib.pyplot as plt

def graficar_potencias():

    x = np.linspace(-10, 10, 300)


    for n in range(1, 11):
        y = x ** n
        plt.plot(x, y, label=f'f(x) = x^{n}')


    plt.axhline(0, color='black', linewidth=0.5)
    plt.axvline(0, color='black', linewidth=0.5)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.title('Gráficas de f(x) = x^n para n ∈ [1, 10]')
    plt.legend()
    plt.show()

# Llamar a la función para graficar
graficar_potencias()
```

```python
import matplotlib.pyplot as plt
import random

def sierpinski_while(iterations):
    vertices = [(0, 0), (1, 0), (0.5, 0.866)]
    x_points, y_points = [], []

    x, y = random.choice(vertices)

    i = 0
    while i < iterations:
        vertex = random.choice(vertices)
        x = (x + vertex[0]) / 2
        y = (y + vertex[1]) / 2
        x_points.append(x)
        y_points.append(y)
        i += 1

    plt.figure(figsize=(6, 6))
    plt.scatter(x_points, y_points, s=0.2, color="black")
    plt.axis("off")
    plt.show()

sierpinski_while(10000)
```
