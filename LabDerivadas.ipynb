```python
import matplotlib.pyplot as plt
import numpy as np
def newton(fun, der, x_n, epsilon=1e-6, steps=50):
  for n in range(steps + 1):
    # Evaluación de la función para ver si el resultado es válido
    f_x = fun(x_n)
    if abs(f_x) < epsilon:
        return x_n
    # Evaluación de la derivada
    d_f = der(x_n)
    if d_f == 0:
        print('Error la derivada es cero')
        return None
    # Estimación del siguiente punto
    x_n = x_n - f_x / d_f
  return None
```

b. $f(2)=-3$ y $f(6)=1$

c. $f'(2)=0$ para $x\neq 2$

d. $f'(6)=3$

```python
import matplotlib.pyplot as plt
import numpy as np
#punto I
#1
#Analisis
# Dominio: Todos los reales excepto x = -1 (denominador = 0)
# Simetría: f(-x) = -x / (-x + 1) ≠ f(x) ⇒ No par ni impar
# Asíntota vertical: x = -1
# Asíntota horizontal: lim x→∞ = 1
def f1(x):
    return x / (x + 1)

def df1(x):
    return 1 / (x + 1)**2

x_vals = np.linspace(-5, 5, 1000)
x_vals = x_vals[np.abs(x_vals + 1) > 0.01]  # evitar asíntota

plt.figure(figsize=(7, 4))
plt.title("f(x) = x / (x + 1)")
plt.axhline(0, color='gray', lw=0.5)
plt.axvline(-1, color='red', linestyle='--', label='Asíntota x = -1')
plt.axhline(1, color='green', linestyle='--', label='Asíntota y = 1')
plt.plot(x_vals, f1(x_vals), label='f(x)')
plt.legend()
plt.grid(True)
plt.show()

#2
# Análisis:
# Para x < 0: g(x) = 0
# Para x ≥ 0: g(x) = x / (2(3x + 2))
# Asíntota vertical: x = -2/3
def g(x):
    return np.where(x < 0, 0, x / (2 * (3*x + 2)))

x_vals = np.linspace(-5, 5, 1000)
x_vals = x_vals[np.abs(x_vals + 2/3) > 0.01]

plt.figure(figsize=(7, 4))
plt.title("g(x) = (|x| + x)/(2(3x + 2))")
plt.axhline(0, color='gray', lw=0.5)
plt.axvline(-2/3, color='red', linestyle='--', label='Asíntota x = -2/3')
plt.plot(x_vals, g(x_vals), label='g(x)')
plt.legend()
plt.grid(True)
plt.show()

#3
#Analisis
# Dominio: x ≠ ±1
# Simetría: Par (f(-x) = f(x))
# Asíntotas verticales: x = ±1
# Asíntota horizontal: y = 1
def f3(x):
    return x**2 / (x**2 - 1)

x_vals = np.linspace(-5, 5, 1000)
x_vals = x_vals[(np.abs(x_vals - 1) > 0.01) & (np.abs(x_vals + 1) > 0.01)]

plt.figure(figsize=(7, 4))
plt.title("f(x) = x² / (x² - 1)")
plt.axhline(1, color='green', linestyle='--', label='Asíntota y = 1')
plt.axvline(1, color='red', linestyle='--')
plt.axvline(-1, color='red', linestyle='--', label='Asíntotas x = ±1')
plt.plot(x_vals, f3(x_vals), label='f(x)')
plt.legend()
plt.grid(True)
plt.show()

#4
#Analisis
# Dominio: x ≠ 3.141 / 7.126 ≈ 0.44
# Asíntota oblicua: División larga ⇒ cociente = 0.734x^2 + ...
def h(x):
    return (5.235 * x**3 - 1.245) / (7.126 * x - 3.141)

x_vals = np.linspace(-5, 5, 1000)
x_vals = x_vals[np.abs(x_vals - (3.141 / 7.126)) > 0.01]

plt.figure(figsize=(7, 4))
plt.title("h(x) = (5.235x³ - 1.245)/(7.126x - 3.141)")
plt.axvline(3.141 / 7.126, color='red', linestyle='--', label='Asíntota x ≈ 0.44')
plt.plot(x_vals, h(x_vals), label='h(x)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Punto II

# f(x) = 2x^2 → f(0)=0, f(1)=2, f es par, derivada positiva para x>0 y segunda derivada positiva
def f(x):
    return 2 * x**2

# Dominio de -2 a 2 para incluir simetría
x = np.linspace(-2, 2, 400)
y = f(x)

# Graficar
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='f(x) = 2x²', color='blue')
plt.axhline(0, color='black', linewidth=0.5)  # eje x
plt.axvline(0, color='black', linewidth=0.5)  # eje y
plt.scatter([0, 1], [f(0), f(1)], color='red', label='Puntos dados')
plt.title("Bosquejo de una función continua, par,\n creciente y cóncava hacia arriba para x>0")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
#Punto III

import numpy as np
import matplotlib.pyplot as plt

# Puntos clave
x_points = np.array([2, 4, 6, 7])
y_points = np.array([-3, -0.5, 1, 1.5])


coefficients = np.polyfit(x_points, y_points, 3)
x_smooth = np.linspace(2, 7, 400)
y_smooth = np.polyval(coefficients, x_smooth)


plt.figure(figsize=(8, 5))
plt.plot(x_smooth, y_smooth, label="Bosquejo de f(x)", color='blue')
plt.plot(x_points, y_points, 'ro', label="Puntos clave")

plt.text(2, -3.3, "(2, -3)", ha='center')
plt.text(6, 1.2, "(6, 1)", ha='center')


plt.title("Bosquejo de una función con condiciones dadas")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
