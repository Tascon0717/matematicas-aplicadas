```python
#EJERCICIO 2
import numpy as np

# Se definen las funciones que representan cada sucesión del ejercicio
def seq_a(n): return 1 / (3**n)
def seq_b(n): return np.pi**n
def seq_c(n): return (3*n-1) / (2*n+6)
def seq_d(n): return np.cos(np.pi*n)
def seq_e(n): return (3*n**4-6*n**3+1) / (5*n**3 + n**2 + 2)
def seq_f(n): return (2*n**5-3*n**2+23) / (7*n**5+13*n**4-12)
def seq_g(n): return (45 - 16*n**2) / (13 + 5*n + 6*n**3)
def seq_h(n): return (3*n + 1) / (np.sqrt(4*n**2 + 1))
def seq_i(n): return (-2)**(2*n + 1)
def seq_j(n): return (10 - 16*n**3) / (1 + n**2)
def seq_k(n):
    expr = (3*n**2 + n - 6) / (5*n**2 + 16)
    return np.sqrt(np.maximum(expr, 0))
def seq_l(n): return (-1)**n / (5**n)

# Diccionario de secuencias
secuencias = {
    "ejercicioA_n": seq_a,
    "ejercicioB_n": seq_b,
    "ejercicioC_n": seq_c,
    "ejercicioD_n": seq_d,
    "ejercicioE_n": seq_e,
    "ejercicioF_n": seq_f,
    "ejercicioG_n": seq_g,
    "ejercicioH_n": seq_h,
    "ejercicioI_n": seq_i,
    "ejercicioJ_n": seq_j,
    "ejercicioK_n": seq_k,
    "ejercicioL_n": seq_l
}

n_valores = np.arange(1, 100)
divergente = []
convergente = {}

# Procesar cada secuencia
for nombre, seq in secuencias.items():
    valores = seq(n_valores)
    limite = np.nanmean(valores[-10:])

    if (np.any(np.abs(valores[-10:]) > 1e6) or  # Crece a infinito
        np.all(np.abs(np.diff(valores[-10:])) > 1) or  # Si tiene un cambio brusco
        np.all(np.isnan(valores[-10:]))):  # Contiene NaN
        divergente.append(nombre)
    else:
        convergente[nombre] = limite

# Imprimir resultados
print("Secuencias convergentes y sus límites")
for nombre, limite in convergente.items():
    print(f"{nombre}: {abs(limite):.4f}")  # Muestra 4 decimales

print("\nSecuencias divergentes")
print(divergente)
```

```python
#Ejercicio 5
import numpy as np
import matplotlib.pyplot as plt

def a_n(n):
    return n * np.sin(1/n)

# a)
a_values = [a_n(n) for n in range(1, 6)]
print("(a) Values of a_n for n = 1 to 5:", a_values)

# b)
n_values = np.arange(1, 201)
a_n_values = a_n(n_values)

# c)
plt.figure(figsize=(8, 5))
plt.plot(n_values, a_n_values, label="$a_n = n \sin(1/n)$", marker='o', linestyle='none', markersize=3)
plt.axhline(y=1, color='r', linestyle='dashed', label="$y = 1$")
plt.xlabel("n")
plt.ylabel("a_n")
plt.legend()
plt.title("Plot of $a_n = n \sin(1/n)$")
plt.show()

# d)
print("Parece razonable que lim a_n = 1 as n -> infinito")

# e)
n_e = 1
while a_n(n_e) <= 0.999:
    n_e += 1
print(f"El menor valor de n para el cual a_n > 0.999 es {n_e}")

# f)
n_f = 1
while abs(a_n(n_f) - 1) >= 0.0001:
    n_f += 1
print(f"El primer valor de n para el cual |a_n - 1| < 0.0001 es {n_f}")
```

```python
#EJERCICIO 8

import numpy as np
import matplotlib.pyplot as plt

# Definir la sucesión
n_values = np.arange(1, 51)
a_values = np.zeros(50)
a_values[0] = 10

# Generar los términos de la sucesión
for i in range(1, 50):
    a_values[i] = (1/3) * a_values[i-1]

# Graficar la sucesión
plt.figure(figsize=(8, 4))
plt.scatter(n_values, a_values, color='blue', label='$a_n = \\frac{1}{3} a_{n-1}$')
plt.plot(n_values, a_values, linestyle='dashed', color='black')
plt.xlabel('$n$')
plt.ylabel('$a_n$')
plt.title('Grafica de la sucesion')
plt.legend()
plt.grid()
plt.show()

# Determinar el límite
limite = np.round(a_values[-1], 6)
print(f"\nEl valor de a_n se aproxima a {limite}, por lo que la sucesion converge a 0.")
```

```python
#EJERCICIO 9
import matplotlib.pyplot as plt
import numpy as np

# Definir la secuencia
def sequence(n):
    return 2**n

# Generar valores de n
n_values = np.arange(0, 10)
a_values = [sequence(n) for n in n_values]

# Graficar la secuencia
plt.figure(figsize=(8, 5))
plt.plot(n_values, a_values, 'bo-', label='$a_n = 2^n$')
plt.xlabel('n')
plt.ylabel('$a_n$')
plt.title('Gráfica de la secuencia $a_n = 2^n$')
plt.legend()
plt.grid()
plt.show()
```
